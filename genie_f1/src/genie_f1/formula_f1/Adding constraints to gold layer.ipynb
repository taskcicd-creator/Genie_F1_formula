# Adding Constraints to Gold Layer

Adds primary key and foreign key constraints to the `formula1_incr.gold` star schema tables.

```python
%sql
-- ============================================================
-- PRIMARY KEY constraints
-- ============================================================

-- ref_nationality_region: lookup table for nationality -> region
ALTER TABLE formula1_incr.gold.ref_nationality_region
  ADD CONSTRAINT pk_ref_nationality_region PRIMARY KEY (nationality);

-- dim_drivers: one row per driver
ALTER TABLE formula1_incr.gold.dim_drivers
  ADD CONSTRAINT pk_dim_drivers PRIMARY KEY (driver_id);

-- dim_constructors: one row per constructor
ALTER TABLE formula1_incr.gold.dim_constructors
  ADD CONSTRAINT pk_dim_constructors PRIMARY KEY (constructor_id);

-- dim_races: one row per (season, round)
ALTER TABLE formula1_incr.gold.dim_races
  ADD CONSTRAINT pk_dim_races PRIMARY KEY (season, round);

-- fact_session_results: one row per driver per session per race
ALTER TABLE formula1_incr.gold.fact_session_results
  ADD CONSTRAINT pk_fact_session_results PRIMARY KEY (season, round, driver_id, session_type);
```

```python
%sql
-- ============================================================
-- FOREIGN KEY constraints
-- ============================================================

-- dim_constructors.nationality -> ref_nationality_region
ALTER TABLE formula1_incr.gold.dim_constructors
  ADD CONSTRAINT fk_constructors_nationality
  FOREIGN KEY (nationality) REFERENCES formula1_incr.gold.ref_nationality_region (nationality);

-- fact_session_results -> dim_races
ALTER TABLE formula1_incr.gold.fact_session_results
  ADD CONSTRAINT fk_fsr_races
  FOREIGN KEY (season, round) REFERENCES formula1_incr.gold.dim_races (season, round);

-- fact_session_results -> dim_drivers
ALTER TABLE formula1_incr.gold.fact_session_results
  ADD CONSTRAINT fk_fsr_drivers
  FOREIGN KEY (driver_id) REFERENCES formula1_incr.gold.dim_drivers (driver_id);

-- fact_session_results -> dim_constructors
ALTER TABLE formula1_incr.gold.fact_session_results
  ADD CONSTRAINT fk_fsr_constructors
  FOREIGN KEY (constructor_id) REFERENCES formula1_incr.gold.dim_constructors (constructor_id);
```
